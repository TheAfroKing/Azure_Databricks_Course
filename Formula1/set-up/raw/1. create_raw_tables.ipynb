## Create circuits table from CSV

```sql
DROP TABLE IF EXISTS f1_raw.circuits;
CREATE TABLE IF NOT EXISTS f1_raw.circuits(
   circuitId INT,
   circuitRef STRING,
   name STRING,
   location STRING,
   country STRING,
   lat DOUBLE,
   lng DOUBLE,
   alt INT,
   URL STRING
)
USING csv
OPTIONS (path 'abfss://raw@f1datalakefp.dfs.core.windows.net/circuits.csv', header true)
```

```sql
SELECT * FROM f1_raw.circuits
```

## Create races table from CSV

```sql
DROP TABLE IF EXISTS f1_raw.races;
CREATE TABLE IF NOT EXISTS f1_raw.races(
  raceId INT,
  year INT,
  round INT,
  circuitId INT,
  name STRING,
  date DATE,
  time STRING,
  url STRING
)
USING csv
OPTIONS (path 'abfss://raw@f1datalakefp.dfs.core.windows.net/races.csv', header true)
```

```sql
SELECT * FROM f1_raw.races
```

## Create constructors table from JSON single line and simple structure 

```sql
DROP TABLE IF EXISTS f1_raw.constructors;
CREATE TABLE IF NOT EXISTS f1_raw.constructors(
  constructorId INT,
  constructorRef STRING,
  name STRING,
  nationality STRING
)
USING json
OPTIONS (path 'abfss://raw@f1datalakefp.dfs.core.windows.net/constructors.json')
```

```sql
select * from f1_raw.constructors
```

## Create drivers table from JSON single line and complex structure

```sql
DROP TABLE IF EXISTS f1_raw.drivers;
CREATE TABLE IF NOT EXISTS f1_raw.drivers(

 driverId INT ,
 driverRef STRING ,
 number INT ,
 code STRING ,
 name STRUCT <forename: STRING , surname: STRING>,
 dob STRING,
 nationality STRING 
)
USING json
OPTIONS (path 'abfss://raw@f1datalakefp.dfs.core.windows.net/drivers.json')
```

```sql
select * from f1_raw.drivers
```

## Create results table from JSON single line and simple structure


```sql
DROP TABLE IF EXISTS f1_raw.results;
CREATE TABLE IF NOT EXISTS f1_raw.results(
  resultId INT,
  raceId INT, 
  driverId INT, 
  constructorId INT,
  number STRING, 
  grid INT, 
  position STRING, 
  positionText STRING, 
  positionOrder INT, 
  points DOUBLE, 
  laps INT, 
  time STRING, 
  milliseconds INT,
  fastestLap STRING, 
  rank INT, 
  fastestLapSpeed STRING, 
  fastestLapTime STRING, 
  statusId INT
)
USING json
OPTIONS (path 'abfss://raw@f1datalakefp.dfs.core.windows.net/results.json')
```

```sql
select * from f1_raw.results
```
