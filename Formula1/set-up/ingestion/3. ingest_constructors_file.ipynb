# Ingest Constructors.json file Assignment


## Assignment
- create a DF loading the constructors.json file in the raw container
- rename columns to (constructor_id,constructor_ref)
- add a new columns ingestion_timestampt
- save the file in parquet file in the processed container.
- verify the schema of the partquet file

```python
from pyspark.sql.functions import current_timestamp,col
```

```python
constructors_schema="constructorId INT,constructorRef STRING,name STRING,nationality STRING"
```

```python
constructors_df = spark.read.schema(constructors_schema).json("abfs://raw@f1datalakefp.dfs.core.windows.net/constructors.json")
```

```python

constructors_df = constructors_df.select(col('constructorId').alias('constructor_id'),col('constructorRef').alias('constructor_ref'),col('name'),col('nationality'))
```

```python
constructors_df = constructors_df.withColumn('ingestion_timestamp', current_timestamp())
```

```python
display(constructors_df)
```

## Write DF into parquet file

```python
constructors_df.write.mode("overwrite").parquet("abfs://processed@f1datalakefp.dfs.core.windows.net/constructors")
```

```python
df=spark.read.parquet("abfs://processed@f1datalakefp.dfs.core.windows.net/constructors")
df.printSchema()
```
