# access dataframes using sql
## Objectives
1. Create temporary views on dataframes
2. Access the view from SQL cell
3. Access the view from Python cell

```python
%run "../includes/configuration"
```

```python
race_results_df= spark.read.parquet(f'{presentation_folder_path}/race_results')
```

```python
race_results_df.createOrReplaceTempView('v_race_results')
```

```python
result = spark.sql("SELECT * FROM v_race_results WHERE race_year = 2020")
```

```python
%sql
SELECT * FROM v_race_results WHERE race_year = 2020
```

# Global temporary view
## Objectives
1. Create temporary views on dataframes
2. Access the view from SQL cell

```python
race_results_df.createOrReplaceGlobalTempView('gv_race_results')
```

```python
%sql 
show tables in global_temp
```

```python
%sql
SELECT * FROM global_temp.gv_race_results WHERE race_year = 2020
```
