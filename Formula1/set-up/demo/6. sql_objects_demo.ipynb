# Learning objectives
1. Create external table using Python
2. Create external table using SQL
3. Effect of droping external table


```sql
%python
race_results_df.write.format("parquet").option("path",f"{presentation_folder_path}/race_results_external_py").saveAsTable("demo.race_results_python_external_")
```

```sql
DESCRIBE EXTENDED demo.race_results_python_external_
```

```sql
DROP TABLE demo.race_results_sql_external
```

```sql
CREATE TABLE demo.race_results_sql_external
(
  race_year INT,
  race_name STRING,
  race_timestamp TIMESTAMP,
  circuit_location STRING,
  driver_name STRING,
  driver_number INT,
  driver_nationality STRING ,
  team STRING,
  grid INT,
  fastest_lap_time STRING,
  race_time STRING,
  points FLOAT,
  position STRING,
  created_date TIMESTAMP
)
USING parquet
LOCATION "abfss://presentation@f1datalakefp.dfs.core.windows.net/race_results_external_sql"
```

```sql
DESCRIBE EXTENDED demo.race_results_sql_external
```

```sql
INSERT INTO demo.race_results_sql_external SELECT * FROM demo.race_results_python
```
