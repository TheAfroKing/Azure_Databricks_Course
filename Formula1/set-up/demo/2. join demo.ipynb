# Outer Joints

## left Outer

```python
circuits_df= spark.read.parquet(f"{processed_folder_path}/circuits").where("circuit_id < 70").withColumnRenamed('name','circuit_name')
```

```python
joined_df = races_df.join(circuits_df, races_df.circuit_id == circuits_df.circuit_id, "left").select(circuits_df.circuit_name, circuits_df.location, circuits_df.country, races_df.race_name,races_df.round)
```

```python
display(joined_df)
```

## **rigth** Outer

```python
joined_df = races_df.join(circuits_df, races_df.circuit_id == circuits_df.circuit_id, "right").select(circuits_df.circuit_name, circuits_df.location, circuits_df.country, races_df.race_name,races_df.round)
```

```python
display(joined_df)
```

## **full** Outer

```python
joined_df = races_df.join(circuits_df, races_df.circuit_id == circuits_df.circuit_id, "full").select(circuits_df.circuit_name, circuits_df.location, circuits_df.country, races_df.race_name,races_df.round)
```

```python
display(joined_df)
```

## Semi joints
like a inner join but only can select the left table columns

```python
joined_df = races_df.join(circuits_df, races_df.circuit_id == circuits_df.circuit_id, "semi").select(races_df.circuit_id, races_df.race_name, races_df.race_id)
```

```python
display(joined_df)
```

## Anti joints
like a inner join but only can select the rigth table columns

```python
joined_df = races_df.join(circuits_df, races_df.circuit_id == circuits_df.circuit_id, "anti")
```

```python
display(joined_df)
```

## Cross joints

```python
joined_df = races_df.crossJoin(circuits_df)
```

```python
display(joined_df)
```
