# Access Azure Data Lake using SAS Token
1. Set the spark config fs azure SAS Token
2. List Files From demo container
3. Read data frin circuits.csv file


```python
dbutils.secrets.help()
```

```python
dbutils.secrets.listScopes()
```

```python
dbutils.secrets.list("formula1")  # Verify Key Vault URL and access permissions
```

```python
dbutils.secrets.get(scope="formula1", key="f1-account-key")
```

```python
spark.conf.set("fs.azure.account.key.f1datalakefp.dfs.core.windows.net", dbutils.secrets.get(scope="formula1", key="f1-account-key"))
```

```python
dbutils.fs.ls("abfss://demo@f1datalakefp.dfs.core.windows.net")
```

```python
display(spark.read.csv("abfs://demo@f1datalakefp.dfs.core.windows.net/circuits.csv"))
```
