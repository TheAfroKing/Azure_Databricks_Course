# Produce the constructors standings assignment

```python
%run "../includes/configuration"
```

```python
race_results_df= spark.read.parquet(f"{presentation_folder_path}/race_results")
```

```python
display(race_results_df)
```

```python
from pyspark.sql.functions import sum, when, col, count, desc ,rank, asc
from pyspark.sql.window import Window
```

```python
constructors_standings_df = race_results_df.groupBy("race_year", "team").agg(sum("points").alias("total_points"), count(when(col("position") == 1, True)).alias("wins"))
```

```python
constructors_rank_spec = Window.partitionBy("race_year").orderBy(desc("total_points"), desc("wins"))
final_df= constructors_standings_df.withColumn("rank", rank().over(constructors_rank_spec))
```

```python
display(final_df.where("race_year = 2020"))
```

```python
final_df.write.mode("overwrite").parquet(f"{presentation_folder_path}/constructors_standings")
```
