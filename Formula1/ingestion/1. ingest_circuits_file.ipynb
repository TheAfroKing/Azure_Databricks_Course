```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType
from pyspark.sql.functions import current_timestamp,to_date,current_date,lit
```

```python
circuits_schema = StructType(fields=[StructField("circuitId", IntegerType(), False),
                                     StructField("circuitRef", StringType(), True),
                                     StructField("name", StringType(), True),
                                     StructField("location", StringType(), True),
                                     StructField("country", StringType(), True,),
                                     StructField("latitude", DoubleType(), True),
                                     StructField("longitude", DoubleType(), True),
                                     StructField("altitude", IntegerType(), True)
                                    ])
circuits_df = spark.read.schema(circuits_schema).csv("abfs://demo@f1datalakefp.dfs.core.windows.net/circuits.csv",header=True)
```

## Add a column
- withColumn function add a column
- current_timeStampt return the current timestamp as a column
- current_date return the current date as a column
- lit converts a literal value to a column

```python
circuits_df = circuits_df.withColumn('current_timestamp', current_timestamp())
#.withColumn('environment', lit("Production")).withColumn('current_date', current_date())
#display(circuits_df)
```

## Write DF into parquet file

```python
circuits_df.write.mode("overwrite").parquet("abfs://processed@f1datalakefp.dfs.core.windows.net/circuits")
```
